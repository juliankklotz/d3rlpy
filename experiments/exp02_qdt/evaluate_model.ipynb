```python
import d3rlpy
from d3rlpy.algos import CQL, IQL
import gym
from d3rlpy.dataset import InfiniteBuffer, ReplayBuffer

replay_buffer = ReplayBuffer
env = gym.make("Hopper-v4")
env_eval = d3rlpy.metrics.EnvironmentEvaluator(env, n_trials=100)
model_path = "/gpfs/data/fs72297/jklotz/programming/cloned_repos/forked_repos_for_master_thesis/d3rlpy/experiments/exp02_qdt/d3rlpy_logs/CQL_Hopper-v4_1_20250710203053/model_300000.d3"
q_algo_loaded = d3rlpy.load_learnable(model_path)
if not isinstance(q_algo_loaded, (CQL, IQL)):
    raise ValueError(
        "The loaded model is not an instance of CQL or IQL."
    )
# cast to the expected type
q_algo = q_algo_loaded
```

```python
mean_score = env_eval(q_algo,replay_buffer)
```
