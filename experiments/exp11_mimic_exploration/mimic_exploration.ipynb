```python
import numpy as np
import pandas as pd

def compute_sofa_scores(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # Respiratory (paO2 / FiO2)
    df["paO2_FiO2"] = df["paO2"] / df["FiO2_1"]
    df["SOFA_resp"] = pd.cut(
        df["paO2_FiO2"],
        bins=[-np.inf, 100, 200, 300, 400, np.inf],
        labels=[4, 3, 2, 1, 0]
    ).astype("float")

    # Coagulation (Platelets)
    df["SOFA_coag"] = pd.cut(
        df["Platelets_count"],
        bins=[-np.inf, 20, 50, 100, 150, np.inf],
        labels=[4, 3, 2, 1, 0]
    ).astype("float")

    # Liver (Bilirubin)
    df["SOFA_liver"] = pd.cut(
        df["Total_bili"],
        bins=[-np.inf, 1.2, 2.0, 6.0, 12.0, np.inf],
        labels=[0, 1, 2, 3, 4]
    ).astype("float")

    # Cardiovascular (MAP or vasopressors proxy)
    conditions = [
        (df["max_dose_vaso"] > 0.1),
        (df["max_dose_vaso"] > 0),
        (df["MeanBP"] < 65),
        (df["MeanBP"] < 70)
    ]
    choices = [4, 3, 2, 1]
    df["SOFA_cardio"] = np.select(conditions, choices, default=0).astype("float")


    # CNS (GCS)
    df["SOFA_cns"] = pd.cut(
        df["GCS"],
        bins=[-np.inf, 5, 9, 12, 14, 15],
        labels=[4, 3, 2, 1, 0],
        include_lowest=True
    ).astype("float")

    # Renal (Creatinine or Urine Output)
    df["SOFA_renal"] = pd.cut(
        df["Creatinine"],
        bins=[-np.inf, 1.2, 2.0, 3.5, 5.0, np.inf],
        labels=[0, 1, 2, 3, 4]
    ).astype("float")

    # Total SOFA
    sofa_cols = [
        "SOFA_resp", "SOFA_coag", "SOFA_liver",
        "SOFA_cardio", "SOFA_cns", "SOFA_renal"
    ]
    df["SOFA_total"] = df[sofa_cols].sum(axis=1, skipna=False)

    return df

mimic_df = compute_sofa_scores(mimic_df)
```

```python
import numpy as np
import pandas as pd


def compute_sofa_scores(
    df: pd.DataFrame,
    timestep_resolution: float = 4.0,          # hours between successive rows
) -> pd.DataFrame:
    """
    SOFA score implementation that mirrors the NumPy logic used by the team.
    Requires the following columns (case-sensitive):
        paO2, FiO2_1, Platelets_count, Total_bili,
        MeanBP, max_dose_vaso, GCS, Creatinine, output_step
    """
    df = df.copy()

    # ───────────────────────── Respiratory ──────────────────────────
    df["paO2_FiO2"] = df["paO2"] / df["FiO2_1"]
    pfi = df["paO2_FiO2"]
    df["SOFA_resp"] = np.select(
        [pfi > 400,
         (pfi >= 300) & (pfi < 400),
         (pfi >= 200) & (pfi < 300),
         (pfi >= 100) & (pfi < 200)],
        [0, 1, 2, 3],
        default=4
    ).astype(float)

    # ──────────────────────── Coagulation ───────────────────────────
    pl = df["Platelets_count"]
    df["SOFA_coag"] = np.select(
        [pl > 150,
         (pl >= 100) & (pl < 150),
         (pl >= 50)  & (pl < 100),
         (pl >= 20)  & (pl < 50)],
        [0, 1, 2, 3],
        default=4
    ).astype(float)

    # ────────────────────────── Liver ───────────────────────────────
    bili = df["Total_bili"]
    df["SOFA_liver"] = np.select(
        [bili < 1.2,
         (bili >= 1.2) & (bili < 2.0),
         (bili >= 2.0) & (bili < 6.0),
         (bili >= 6.0) & (bili < 12.0)],
        [0, 1, 2, 3],
        default=4
    ).astype(float)

    # ────────────────────── Cardiovascular ──────────────────────────
    map_  = df["MeanBP"]
    vaso  = df["max_dose_vaso"]
    df["SOFA_cardio"] = np.select(
        [map_ >= 70,
         (map_ < 70) & (map_ >= 65),
         map_ < 65,
         (vaso > 0)  & (vaso <= 0.1),
         vaso > 0.1],
        [0, 1, 2, 3, 4]
    ).astype(float)

    # ─────────────────────────── CNS ────────────────────────────────
    gcs = df["GCS"]
    df["SOFA_cns"] = np.select(
        [gcs > 14,
         (gcs > 12) & (gcs <= 14),
         (gcs >  9) & (gcs <= 12),
         (gcs >  5) & (gcs <=  9)],
        [0, 1, 2, 3],
        default=4
    ).astype(float)

    # ────────────────────────── Renal ───────────────────────────────
    cr   = df["Creatinine"]
    uo   = df["output_step"]                        # mL in this timestep
    thr3 = 500 * timestep_resolution / 24          # e.g. 83.33 mL for 4-hour step
    thr4 = 200 * timestep_resolution / 24          # e.g. 33.33 mL for 4-hour step

    df["SOFA_renal"] = np.select(
        [cr < 1.2,
         (cr >= 1.2) & (cr < 2.0),
         (cr >= 2.0) & (cr < 3.5),
         ((cr >= 3.5) & (cr < 5.0)) | (uo < thr3),
         (cr > 5.0)               | (uo < thr4)],
        [0, 1, 2, 3, 4]
    ).astype(float)

    # ──────────────────────── Total score ───────────────────────────
    sofa_cols = [
        "SOFA_resp", "SOFA_coag", "SOFA_liver",
        "SOFA_cardio", "SOFA_cns", "SOFA_renal"
    ]
    df["SOFA_total"] = df[sofa_cols].sum(axis=1, skipna=False)

    return df
mimic_df = compute_sofa_scores(mimic_df)
```

```python
import pandas as pd
import numpy as np

def pad_fast(df: pd.DataFrame,
             freq: str = "4h",
             id_col: str = "icustayid",
             time_col: str = "timestep_dt",
             action_cols: list[str] | None = None,
             ffill_cols:  list[str] | None = None,
             reward_col:  str | None = None) -> pd.DataFrame:
    """
    Vectorised, ~10× faster version of pad_to_regular_grid.
    Requires pandas ≥ 2.3 (for GroupBy.resample).
    """
    action_cols = action_cols or []
    ffill_cols  = ffill_cols  or []

    df = df.copy()
    df["_orig"] = 1                    # sentinel to spot real rows
    df = df.sort_values([id_col, time_col]).set_index(time_col)

    # --- 1. build the regular grid & mark synthetic rows --------------------
    padded = (df
              .groupby(id_col)         # stays stay separate, C-level loop
              .resample(freq)          # up-sample index (adds NaNs)
              .first()                 # keep existing rows where they exist
    )
    padded["is_padding"] = padded["_orig"].isna()

    # --- 2. restore keys & fill state --------------------------------------
    padded[id_col] = padded[id_col].ffill()  # cheap because already sorted
    if ffill_cols:
        padded[ffill_cols] = padded[ffill_cols].ffill().infer_objects(copy=False)

    for c in action_cols:                        # zero-out actions on gaps
        padded.loc[padded["is_padding"], c] = 0

    if reward_col:
        padded.loc[padded["is_padding"], reward_col] = 0.0

    # elapsed time since previous step (optional but often handy)
    idx_time = padded.index.get_level_values(time_col)
    padded["delta_hours"] = (
        idx_time
        .to_series(index=padded.index)           # align with MultiIndex
        .groupby(padded[id_col]).diff()
        .dt.total_seconds().div(3600)
    )

    # --- 3. flatten MultiIndex & tidy up -----------------------------------
    padded = (padded
              .reset_index(level=1)             # drop the resample level
              .rename(columns={time_col: time_col})
              .reset_index(drop=True)
              .drop(columns="_orig")
              .sort_values([id_col, time_col])
    )
    return padded

```

```python
import numpy as np
import pandas as pd
from d3rlpy.dataset import Episode, ReplayBuffer, FIFOBuffer 
from d3rlpy.constants import ActionSpace

def dataframe_to_replaybuffer(df, obs_cols, action_cols, reward_col,
                              time_col="timestep_dt", id_col="icustayid"):
    
    df = df.sort_values([id_col, time_col])

    episodes = []
    for stay_id, g in df.groupby(id_col, sort=False):
        obs  = g[obs_cols].to_numpy(dtype=np.float32)
        acts = g[action_cols].to_numpy(dtype=np.float32)    #  (T, 2)
        rews = g[reward_col].to_numpy(dtype=np.float32)

        # terminals         = np.zeros(len(g), bool)
        # terminals[-1]     = True

        ep = Episode(
            observations=obs,
            actions=acts,
            rewards=rews,
            terminated=True,
        )
        episodes.append(ep)

    buf = ReplayBuffer(
        buffer=FIFOBuffer(limit=sum(len(ep) for ep in episodes)),
        episodes=episodes,
        action_space=ActionSpace.CONTINUOUS,
        action_size=2,
    )   # action_size=2 is automatically recorded :contentReference[oaicite:1]{index=1}
    return buf

sepsis_continous_buffer = dataframe_to_replaybuffer(sepsis_fast_padded,observation_columns,action_cols=["output_step", "median_dose_vaso"],reward_col="SOFA")
```

```python
import numpy as np
import pandas as pd

def add_action_bin(original_df: pd.DataFrame) -> pd.DataFrame:
    """
    Add a single 'action_bin' column to df by binning fluid_step_ml and
    vaso_rate_mcgkgm into 5 bins each and computing fluid_bin*5 + vaso_bin.
    Intermediate bins aren’t stored in df.
    """
    df = original_df.copy()
    
    fluid_bins = [-np.inf, 0, 50.00, 152.14, 500.00, np.inf]
    vaso_bins  = [-np.inf, 0, 0.08,   0.20,   0.45,   np.inf]
    labels     = [0, 1, 2, 3, 4]

    df['action_bin'] = (
        pd.cut(
            df['output_step'].fillna(0),
            bins=fluid_bins,
            labels=labels,
            right=True
        ).astype('int8') * 5
        + pd.cut(
            df['median_dose_vaso'].fillna(0),
            bins=vaso_bins,
            labels=labels,
            right=True
        ).astype('int8')
    ).astype('int8')
    
    return df

def add_action_bin_np(original_df: pd.DataFrame) -> pd.DataFrame:
    """
    Add a single 'action_bin' column to df by:
      - binning fluid_step_ml into 5 right‐inclusive buckets [0,50,152.14,500]
      - binning vaso_rate_mcgkgm into 5 right‐inclusive buckets [0,0.08,0.20,0.45]
      - computing action_bin = fluid_bin*5 + vaso_bin
    Treats NaN as zero, and result is cast to int8.
    """

    df = original_df.copy()
    # define the positive cut‐points; values ≤0 map to bin 0, then (0,…]→1…4
    f_edges = [0,   50.00, 152.14, 500.00]
    v_edges = [0,    0.08,   0.20,   0.45]

    df['action_bin'] = (
        5 * np.digitize(df['output_step'].fillna(0).values, f_edges, right=True)
        +     np.digitize(df['median_dose_vaso'].fillna(0).values, v_edges, right=True)
    ).astype('int8')

    return df
```

```python
import numpy as np
import pandas as pd
from d3rlpy.dataset import Episode, ReplayBuffer, FIFOBuffer 
from d3rlpy.constants import ActionSpace

def dataframe_to_replaybuffer(df, obs_cols, action_cols, reward_col,action_space:ActionSpace,action_size:int,
                              time_col="timestep_dt", id_col="icustayid"):
    
    df = df.sort_values([id_col, time_col])

    episodes = []
    for stay_id, g in df.groupby(id_col, sort=False):
        obs  = g[obs_cols].to_numpy(dtype=np.float32)
        acts = g[action_cols].to_numpy(dtype=np.float32)    #  (T, 2)
        rews = g[reward_col].to_numpy(dtype=np.float32)

        terminals         = np.zeros(len(g), bool)
        terminals[-1]     = True

        ep = Episode(
            observations=obs,
            actions=acts,
            rewards=rews,
            terminated=True,
        )
        episodes.append(ep)

    buf = ReplayBuffer(
        buffer=FIFOBuffer(limit=sum(len(ep) for ep in episodes)),
        episodes=episodes,
        action_space=action_space,
        action_size=action_size,
    )   # action_size=2 is automatically recorded :contentReference[oaicite:1]{index=1}
    return buf

sepsis_discrete_buffer = dataframe_to_replaybuffer(sepsis_discrete_actions,observation_columns,action_cols=["action_bin"],reward_col="SOFA",
                                                    action_space=ActionSpace.DISCRETE,action_size=25)
```

```python
from typing import Iterator
import numpy as np
from d3rlpy.dataset import EpisodeBase, TrajectoryMiniBatch
from d3rlpy.dataset.components import PartialTrajectory

def _compute_episode_targeted_rtg(
    rewards: np.ndarray,     # (T, 1) float32
    target_return: float,
    gamma: float = 1.0,
) -> np.ndarray:
    """Compute global targeted RTG G_t for the entire episode.
       Recurrence:
         G_0 = target_return
         G_{t+1} = (G_t - r_t) / gamma
       If gamma == 1.0, this is just cumulative subtraction.
    """
    T = rewards.shape[0]
    G = np.zeros((T, 1), dtype=np.float32)
    g = float(target_return)
    for t in range(T):
        G[t, 0] = g
        if t < T - 1:
            if gamma == 1.0:
                g = g - float(rewards[t, 0])
            else:
                g = (g - float(rewards[t, 0])) / gamma
    return G

def make_targeted_trajectory_batches_global(
    episode: EpisodeBase,
    window_size: int,
    target_return: float,
    gamma: float = 1.0,
) -> Iterator[TrajectoryMiniBatch]:
    """
    Yield TrajectoryMiniBatch windows where returns_to_go come from a SINGLE
    episode-global targeted RTG sequence G (no per-window reset).
    - Left-pads each window to `window_size`.
    - Masks valid steps (1.0) and padding (0.0).
    - Keeps temporal order identical to the episode.
    """
    # assumes ndarray observations; if you have tuple observations, pad each element separately
    assert isinstance(episode.observations, np.ndarray), "Tuple obs not handled here."

    T = episode.transition_count
    n_batches = T // window_size + (1 if T % window_size != 0 else 0)

    obs_full  = episode.observations.astype(np.float32)                    # (T, ...)
    acts_full = episode.actions.astype(np.float32)                         # (T, ...)
    rews_full = episode.rewards.reshape(-1, 1).astype(np.float32)          # (T, 1)
    terms_full = np.zeros((T, 1), dtype=np.float32)
    terms_full[-1, 0] = float(episode.terminated)

    # compute episode-global G once
    G_full = _compute_episode_targeted_rtg(rews_full, target_return, gamma)  # (T,1)

    for i in range(n_batches):
        head = i * window_size
        tail = min(head + window_size, T)

        obs   = obs_full[head:tail]
        acts  = acts_full[head:tail]
        rews  = rews_full[head:tail]
        terms = terms_full[head:tail]
        rtg   = G_full[head:tail]  # <- global targeted RTG slice

        L = tail - head
        pad = window_size - L

        def _pad_left(x: np.ndarray) -> np.ndarray:
            if pad == 0:
                return x
            return np.concatenate([np.zeros((pad,) + x.shape[1:], dtype=np.float32), x], axis=0)

        observations   = _pad_left(obs)
        actions        = _pad_left(acts)
        rewards        = _pad_left(rews)
        returns_to_go  = _pad_left(rtg)
        terminals      = _pad_left(terms)

        timesteps = np.arange(window_size, dtype=np.float32)
        masks = np.zeros((window_size,), dtype=np.float32)
        masks[-L:] = 1.0

        partial = PartialTrajectory(
            observations=observations,
            actions=actions,
            rewards=rewards,
            returns_to_go=returns_to_go,
            terminals=terminals,
            timesteps=timesteps,
            masks=masks,
            length=window_size,
        )
        yield TrajectoryMiniBatch.from_partial_trajectories([partial])
```
